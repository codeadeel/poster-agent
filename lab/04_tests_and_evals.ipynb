# 4. Tests and evals

The 26 tests in `poster-mcp/tests` talk to the MCP server through a real MCP client and use a fake clock, so 5 to 20 second jobs finish instantly. They cover the schemas, validation, the job lifecycle, idempotency, timeouts and limits, and secrets.

```python
import subprocess

def run(command: list[str]) -> None:
    """Run a command in the project folder and print what it prints."""
    result = subprocess.run(command, cwd="/app/poster-mcp", capture_output=True, text=True)
    print(result.stdout.strip())
    if result.returncode != 0:
        print("(exit code", result.returncode, ")")
        print(result.stderr[-1500:])

run(["python", "-m", "pytest", "-v", "--no-header", "-p", "no:cacheprovider"])
```
